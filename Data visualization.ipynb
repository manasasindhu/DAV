```python
# Import Pandas for handling data
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for plotting graphs
import matplotlib.pyplot as plt

# Import Seaborn for statistical visualization
import seaborn as sns
```

```python

# Read the Iris CSV file
df = pd.read_csv("iris.csv")

# Display the first five rows
print(df.head())
```

```python
# Display all column names
print(df.columns)
```

```python
# Plot the distribution of petal length
plt.hist(df["petal_length"], bins=5)

# Add the graph title
plt.title("Distribution of Petal Length")

# Label the X-axis
plt.xlabel("Petal Length (cm)")

# Label the Y-axis
plt.ylabel("Frequency")

# Display the graph
plt.show()
```

```python
# Plot the distribution of petal width
sns.histplot(data=df, x="petal_length", bins=15)

# Add the graph title
plt.title("Distribution of Petal Width")

# Label the X-axis
plt.xlabel("Petal Width (cm)")

# Label the Y-axis
plt.ylabel("Frequency")

# Display the graph
plt.show()
```

```python
# Create a box plot of petal length
plt.boxplot(df["petal_length"])

# Add the graph title
plt.title("Petal Length Box Plot")

# Label the Y-axis
plt.ylabel("Length (cm)")

# Display the graph
plt.show()
```

```python
# Create a box plot of petal width
sns.boxplot(data=df, y="petal_length")

# Add the graph title
plt.title("Petal Width Box Plot")

# Display the graph
plt.show()
```

```python

# Compare petal lengths of different Iris species
sns.boxplot(data=df, x="species", y="petal_length")

# Add the graph title
plt.title("Petal Length by Iris Species")

# Display the graph
plt.show()
```

```python
# Plot petal length against petal width
plt.scatter(df["petal_length"], df["petal_width"])

# Add the graph title
plt.title("Petal Length vs Petal Width")

# Label the X-axis
plt.xlabel("Petal Length (cm)")

# Label the Y-axis
plt.ylabel("Petal Width (cm)")

# Display the graph
plt.show()
```

```python

# Compare petal length and width for each species
sns.scatterplot(
    data=df,
    x="petal_length",
    y="petal_width",
    hue="species"
)

# Add the graph title
plt.title("Petal Length vs Petal Width by Species")

# Display the graph
plt.show()
```

```python
# Compare relationships between numerical features
sns.pairplot(df.drop(columns=["petal_length"]), hue="species")

# Display the graphs
plt.show()
```

```python

# Select numerical columns and exclude the ID column
numerical_df = df.select_dtypes("number").drop(
    columns=["Id"], errors="ignore"
)

# Calculate correlation between numerical features
correlation = numerical_df.corr()

# Display the correlation heatmap
sns.heatmap(correlation, annot=True, cmap="coolwarm")

# Add the graph title
plt.title("Correlation Heatmap")

# Display the graph
plt.show()
```
