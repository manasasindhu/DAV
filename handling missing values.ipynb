```python

# Create a library dataset with missing values
data = {
    'Book': ['Python', 'Java', 'C Programming', 'SQL', 'Python', 'Java'],
    'Price': [400, 500, None, 350, 400, None],
    'Rating': [4, 5, 3, None, 4, 5],
    'Category': ['Programming', 'Programming', None, 'Database', 'Programming', 'Programming'],
    'Publisher': ['ABC', 'XYZ', 'ABC', None, 'ABC', 'XYZ']
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python

# Create library data with missing categories
data = {
    'Book': ['Python', 'Java', 'C Programming', 'SQL', 'HTML'],
    'Category': ['Programming', 'Programming', None, 'Database', None]
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)
```

```python
# Display data before forward fill
print("Before Forward Fill:")
print(df)
```

```python
# Fill missing categories with the previous available category
df['Category'] = df['Category'].ffill()

# Display data after forward fill
print("\nAfter Forward Fill:")
print(df)
```

```python

# Import LabelEncoder for categorical encoding
from sklearn.preprocessing import LabelEncoder
```

```python

# Create library category data
data = {
    'Book': ['Python', 'Java', 'C Programming', 'SQL'],
    'Category': ['Programming', 'Programming', 'Programming', 'Database']
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# Create a LabelEncoder object
encoder = LabelEncoder()

# Convert categories into numerical labels
df['Category_Encoded'] = encoder.fit_transform(df['Category'])

# Display the encoded dataset
print(df)
```

```python

# Display the category names used by the encoder
print(encoder.classes_)
```

```python
# Import pandas library
import pandas as pd

# Create library category data
data = {
    'Book': ['Python', 'Java', 'C Programming', 'SQL'],
    'Category': ['Programming', 'Programming', 'Programming', 'Database']
}

# Convert dictionary into a DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# Create separate columns for each category
encoded_df = pd.get_dummies(df, columns=['Category'])

# Display the encoded dataset
print(encoded_df)
```

```python
# Convert categories into integer columns containing 0 and 1
encoded_df = pd.get_dummies(
    df,
    columns=['Category'],
    dtype=int
)

# Display the encoded dataset
print(encoded_df)
```

```python
# Import pandas
import pandas as pd

# Create library dataset with missing values
data = {
    'Book': ['Python', 'Java', 'C Programming', 'SQL', 'Python', 'Java'],
    'Price': [400, 500, None, 350, 400, None],
    'Rating': [4, 5, 3, None, 4, 5],
    'Category': ['Programming', 'Programming', None, 'Database', 'Programming', 'Programming'],
    'Publisher': ['ABC', 'XYZ', 'ABC', None, 'ABC', 'XYZ']
}

# Create DataFrame
df = pd.DataFrame(data)

# Display the dataset
print(df)
```

```python
# Check missing values in the Price column
print(df['Price'].isnull().sum())
```

```python
# Replace missing Publisher values with Unknown
df['Publisher'] = df['Publisher'].fillna('Unknown')

# Display the updated Publisher column
print(df['Publisher'])
```

```python

# Remove rows that contain missing values
df_clean = df.dropna()

# Display the cleaned dataset
print(df_clean)
```

```python
# Remove columns that contain at least one missing value
df_no_missing_columns = df.dropna(axis=1)

# Display the result
print(df_no_missing_columns)
```
